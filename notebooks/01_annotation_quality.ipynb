## Reliability and adjudication

Agreement is measured between two independent human annotators. Every semantic
disagreement and inconsistent duplicate-text label enters an adjudication
queue, and downstream analysis remains blocked until final decisions exist.


```python
from phasma_review.annotations import annotation_status, build_adjudication_queue

status = annotation_status()
status
```

```python
# Run only when both status entries report valid_and_complete=True.
if all(status[name].get("valid_and_complete", False) for name in ("annotator_a", "annotator_b")):
    adjudication_queue = build_adjudication_queue()
    display(adjudication_queue.head())
else:
    print("BLOCKED: finish both independent annotation files first.")
```
